# Fermat's Small Theorem

```python
p = 13 # any prime
a = 5

ms = list(range(1, p))
zs = [(a*m) % p for m in ms]
```

```python
for i in range(len(zs)):
    print(f'{ms[i]: <2} * a % p == {zs[i]}')
```

```python
M = 1
for m in ms:
    M *= m

Z = 1
for z in zs:
    Z *= z

A = a**(p-1)

(M * A) % p == Z % p
```

### Coprimes

```python
from math import *

def coprimes(p):
    output = []
    for x in range(1, p):
        if gcd(x, p) == 1:
            output.append(x)
    return output

def phi(n):
    return len(coprimes(n))
```

For any prime number P its coprimes are all number from 1 to P - 1\
And phi is just count of those coprimes

```python
p = 8 # any number, not only primes

cps = coprimes(p)
print(cps)

any_coprime = cps[3]
print([any_coprime * cp % p for cp in cps]) # just shuffles cps
```
